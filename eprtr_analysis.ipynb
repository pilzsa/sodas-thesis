```python
def classify_by_parentheses(name):
    """Check if the name ends with (ABC) pattern"""
    # Match any characters ending with (content) at the end of string
    if re.search(r'\([^)]+\)$', str(name)):
        return 'group'
    else:
        return 'single'

class_sheet['type'] = class_sheet['name'].apply(classify_by_parentheses)

class_sheet.head(2)
```

## Adding EC and CAS for Single Substances to the EPRTR

```python
cas_ec_class = (
    pd.DataFrame.from_dict(
        eprtr_ec_cas_dict_single,
        orient="index"
    )
    .rename_axis("pollutant")
    .reset_index()
)
```

```python
cas_ec_class.head()
```

```python
eprtr_releases_with_cas = eprtr_releases.merge(
    cas_ec_class[
        ["pollutant", "ec", "cas", "group"]
    ],
    how="left",
    left_on="Pollutant",
    right_on="pollutant",
    validate="many_to_one",
    indicator=False,
)
```

#### Adding Boolean Identifier for pollutants, column 'group': True - False, Substance Group - Single Substance

```python
eprtr_releases_with_cas = (
    eprtr_releases_with_cas
    .drop(columns="pollutant")
    .assign(
        group=lambda df: (
            df["group"]
            .fillna(True)
            .astype(bool)
        )
    )
)
```

```python
# Ensure reporting years have the same dtype.
for table in [releases, ied]:
    table["reportingYear"] = (
        pd.to_numeric(table["reportingYear"], errors="raise")
        .astype("Int64")
    )

# Clean identifiers without changing the source tables.
releases["FacilityInspireId"] = (
    releases["FacilityInspireId"]
    .astype("string")
    .str.strip()
    .replace("", pd.NA)
)

ied["parent_facilityInspireId"] = (
    ied["parent_facilityInspireId"]
    .astype("string")
    .str.strip()
    .replace("", pd.NA)
)

ied["InstallationInspireId"] = (
    ied["InstallationInspireId"]
    .astype("string")
    .str.strip()
    .replace("", pd.NA)
)

keys = ["parent_facilityInspireId", "reportingYear"]

# Keep records with incomplete IED join keys separately for auditing.
ied_missing_keys = ied.loc[
    ied[keys].isna().any(axis=1)
].copy()

ied_valid = ied.dropna(subset=keys).copy()
```

```python
ied_missing_keys.head(2)
ied_valid.head(2)
```

```python
parent_year_summary = (
    ied_valid
    .groupby(keys, as_index=False)
    .agg(
        n_ied_rows=("InstallationInspireId", "size"),
        n_ied_installations=("InstallationInspireId", "nunique"),
        n_ied_rows_missing_installation_id=(
            "InstallationInspireId",
            lambda s: s.isna().sum(),
        ),
    )
)

multiple_installations = parent_year_summary.loc[
    parent_year_summary["n_ied_installations"] > 1
].copy()
```

```python
yearly_counts = (
    parent_year_summary
    .assign(
        has_multiple_installations=lambda d: (
            d["n_ied_installations"] > 1
        )
    )
    .groupby("reportingYear", as_index=False)
    .agg(
        n_parent_facilities=("parent_facilityInspireId", "nunique"),
        n_parents_with_multiple_installations=(
            "has_multiple_installations",
            "sum",
        ),
    )
)

display(yearly_counts)

print(
    "Facility-year combinations with multiple installations:",
    len(multiple_installations),
)

print(
    "Distinct facilities with multiple installations in at least one year:",
    multiple_installations["parent_facilityInspireId"].nunique(),
)
```

```python
def distinct_values(series):
    return series.dropna().drop_duplicates().tolist()

context_columns = [
    column
    for column in ied_valid.columns
    if column not in keys
]

ied_by_facility_year = (
    ied_valid
    .groupby(keys, as_index=False)
    .agg({
        column: distinct_values
        for column in context_columns
    })
    .rename(columns={
        column: f"{column}_ied"
        for column in context_columns
    })
    .merge(
        parent_year_summary,
        on=keys,
        how="left",
        validate="one_to_one",
    )
)

assert not ied_by_facility_year.duplicated(keys).any()
```
